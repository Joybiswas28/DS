# 3. Basic graph vocabulary

Before creating graphs, understand these terms:

- **Dataset:** A collection of related values.
- **Observation:** One row or one item being studied.
- **Variable:** A characteristic measured for each observation.
- **X-axis:** The horizontal axis.
- **Y-axis:** The vertical axis.
- **Data point:** One plotted observation. In a scatter plot, one dot usually represents one row.
- **Scale:** The numerical spacing shown along an axis.
- **Tick:** A small mark and label along an axis.
- **Title:** A short description of the graph.
- **Legend:** A key explaining different lines, bars, or groups.
- **Label:** Text explaining an axis or category.

# 4. Histogram: Distribution of exam scores

A histogram shows how often numerical values fall inside ranges called **bins**.

We will use scores from 20 students.

```python
exam_scores = [45, 52, 55, 58, 60, 61, 62, 64, 65, 67,
               68, 70, 72, 74, 75, 78, 80, 82, 85, 90]

scores_df = pd.DataFrame({"Exam_Score": exam_scores})
scores_df
```

### Code explanation

- `exam_scores = [...]` creates a Python **list**.
- Square brackets `[]` are used to store several values in one list.
- Commas separate the values.
- `scores_df` is a variable name.
- `pd.DataFrame(...)` converts the data into a Pandas table.
- `{"Exam_Score": exam_scores}` is a dictionary:
  - Curly brackets `{}` create the dictionary.
  - `"Exam_Score"` becomes the column name.
  - The colon `:` connects the column name to its values.
- `display(scores_df)` shows the complete table neatly.
- Every row represents one student.
- `Exam_Score` is the variable being measured.

```python
plt.figure(figsize=(8, 5))
plt.hist(scores_df["Exam_Score"], bins=5, edgecolor="red")
plt.title("Distribution of Student Exam Scores")
plt.xlabel("Exam score")
plt.ylabel("Number of students")
plt.grid(axis="y", alpha=0.2)
plt.tight_layout()
plt.show()
```

### How to read this histogram

- **X-axis:** Exam-score ranges.
- **Y-axis:** Number of students whose scores fall within each range.
- **One bar:** A range of scores, not one individual student.
- **Bar height:** How many students are inside that score range.
- **A bin:** One interval used to group numerical values.
- The tallest bar shows the score range containing the largest number of students.
- A histogram helps us see the centre, spread, concentration, and possible gaps in numerical data.

### Line-by-line code explanation

- `plt.figure(figsize=(8, 5))`
  - Creates a new graph area.
  - `figsize=(8, 5)` sets its width and height in inches.
  - `(8, 5)` is a tuple containing two values.
- `plt.hist(...)`
  - Creates a histogram.
  - `scores_df["Exam_Score"]` selects the `Exam_Score` column.
  - Square brackets with a column name select that column.
  - `bins=5` divides the score range into five groups.
  - `edgecolor="black"` draws a visible border around each bar.
- `plt.title(...)` adds the graph title.
- `plt.xlabel(...)` labels the horizontal axis.
- `plt.ylabel(...)` labels the vertical axis.
- `plt.grid(axis="y", alpha=0.3)` adds light horizontal guide lines.
  - `axis="y"` means only the y-axis grid is shown.
  - `alpha=0.3` makes the grid partly transparent.
- `plt.tight_layout()` adjusts spacing so labels are not cut off.
- `plt.show()` displays the completed graph.

### When to use a histogram

Use it for the distribution of one numerical variable, such as age, height, marks, income, or temperature.

Do not use it for category names such as departments or fruit types.

# 5. Vertical bar chart: Books borrowed by category

A bar chart compares separate categories.

```python
book_data = pd.DataFrame({
    "Category": ["Science", "History", "Literature", "Technology", "Art"],
    "Books_Borrowed": [32, 18, 27, 40, 15]
})

display(book_data)
```

### Code explanation

- `book_data` stores a table in a variable.
- `pd.DataFrame({...})` creates a DataFrame from a dictionary.
- `"Category"` contains text labels.
- `"Books_Borrowed"` contains numerical counts.
- Each row represents one book category.
- Both lists contain five items, so Pandas can match them row by row.

```python
plt.figure(figsize=(8, 5))
plt.bar(book_data["Category"], book_data["Books_Borrowed"])
plt.title("Books Borrowed from a Library")
plt.xlabel("Book category")
plt.ylabel("Number of books borrowed")
plt.tight_layout()
plt.show()
```

### How to read this bar chart

- **X-axis:** Book categories.
- **Y-axis:** Number of books borrowed.
- **One bar:** One category.
- **Bar height:** The value belonging to that category.
- The tallest bar represents the most frequently borrowed category.
- The shortest bar represents the least frequently borrowed category.

### Line-by-line code explanation

- `plt.bar(x_values, y_values)` creates a vertical bar chart.
- `book_data["Category"]` supplies the category names for the x-axis.
- `book_data["Books_Borrowed"]` supplies the numerical bar heights.
- The first argument controls horizontal positions.
- The second argument controls vertical values.
- The title and axis-label functions work in the same way as before.

### Bar chart versus histogram

- A **bar chart** compares separate categories.
- A **histogram** groups continuous numerical values into ranges.
- Bar-chart bars usually have spaces between them.
- Histogram bars usually touch because the ranges are connected.

# 6. Horizontal bar chart: Weekly hours spent on activities

A horizontal bar chart is useful when category names are long or when there are many categories.

```python
activity_data = pd.DataFrame({
    "Activity": [
        "Attending classes",
        "Independent study",
        "Exercise",
        "Entertainment",
        "Household work",
        "Travelling"
    ],
    "Hours": [20, 14, 5, 10, 7, 8]
})

display(activity_data)
```

### Code explanation

- The DataFrame contains two variables: `Activity` and `Hours`.
- Each row represents one type of weekly activity.
- Text values are written inside quotation marks.
- Values are arranged in the same order in both lists.
- For example, `20` belongs to `"Attending classes"`.

```python
plt.figure(figsize=(9, 5))
plt.barh(activity_data["Activity"], activity_data["Hours"])
plt.title("Weekly Time Spent on Different Activities")
plt.xlabel("Hours per week")
plt.ylabel("Activity")
plt.tight_layout()
plt.show()
```

### How to read this horizontal bar chart

- **X-axis:** Number of hours per week.
- **Y-axis:** Activity names.
- **One bar:** One activity.
- **Bar length:** Hours spent on the activity.
- A longer bar means more time.
- A shorter bar means less time.

### Code explanation

- `plt.barh(...)` creates a horizontal bar chart.
- The letter `h` in `barh` means **horizontal**.
- The first argument contains the category names shown on the y-axis.
- The second argument contains the numerical values shown on the x-axis.
- The remaining lines add a title, labels, spacing, and display.

### When to use it

Use a horizontal bar chart when labels are long, because horizontal space makes the labels easier to read.

# 7. Pie chart: Monthly household spending

A pie chart shows how a total is divided into parts.

The values should represent parts of one meaningful whole.

```python
expense_data = pd.DataFrame({
    "Expense": ["Rent", "Food", "Transport", "Education", "Other"],
    "Amount": [12000, 8000, 3000, 4000, 3000]
})

display(expense_data)

total_expense = expense_data["Amount"].sum()
print("Total monthly expense:", total_expense)
```

### Code explanation

- Each row represents one expense category.
- `expense_data["Amount"]` selects the numerical column.
- `.sum()` adds every value in the selected column.
- The dot before `sum()` means that the method is applied to the selected column.
- `total_expense` stores the result.
- The comma inside `print()` separates the text from the calculated value.

```python
plt.figure(figsize=(7, 7))
plt.pie(
    expense_data["Amount"],
    labels=expense_data["Expense"],
    autopct="%1.1f%%",
    startangle=90
)
plt.title("Monthly Household Spending")
plt.tight_layout()
plt.show()
```

### How to read this pie chart

- A pie chart does not use a normal x-axis or y-axis.
- **One slice:** One expense category.
- **Slice size:** The category's share of total spending.
- A larger slice represents a larger share.
- `Rent` occupies the largest share in this example.
- All slices together represent 100% of monthly spending.

### Line-by-line code explanation

- `plt.pie(...)` creates the pie chart.
- `expense_data["Amount"]` provides the slice sizes.
- `labels=expense_data["Expense"]` writes category names beside slices.
- `autopct="%1.1f%%"` prints percentages:
  - `%` begins the formatting instruction.
  - `1.1f` means show one digit after the decimal point.
  - `%%` displays an actual percent sign.
- `startangle=90` rotates the starting position of the first slice.
- The function call is written on several lines for readability.
- The opening and closing parentheses show that all these lines belong to the same function call.

### When not to use a pie chart

Avoid it when:

- There are many categories.
- Values are not parts of one total.
- Exact comparisons are important.
- You are showing change over time.

# 8. Scatter plot: Study hours and exam marks

A scatter plot shows the relationship between two numerical variables.

Each dot normally represents one observation.

```python
study_data = pd.DataFrame({
    "Study_Hours": [1, 2, 2.5, 3, 4, 4.5, 5, 6, 7, 8],
    "Exam_Mark": [42, 48, 50, 55, 61, 64, 68, 75, 81, 88]
})

display(study_data)
```

### Code explanation

- Every row represents one student.
- `Study_Hours` is one numerical variable.
- `Exam_Mark` is another numerical variable.
- A student's two values belong together in the same row.
- For example, the first student studied for 1 hour and received 42 marks.
- Decimal numbers such as `2.5` and `4.5` are called floating-point numbers.

```python
plt.figure(figsize=(8, 5))
plt.scatter(study_data["Study_Hours"], study_data["Exam_Mark"])
plt.title("Study Hours and Exam Marks")
plt.xlabel("Study hours")
plt.ylabel("Exam mark")
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

### How to read this scatter plot

- **X-axis:** Number of study hours.
- **Y-axis:** Exam mark.
- **One dot:** One student.
- The horizontal position of a dot gives that student's study hours.
- The vertical position gives that student's mark.
- A dot near `(4, 61)` represents a student who studied for about 4 hours and scored about 61.
- Most dots move upward from left to right.
- This suggests a **positive relationship**: students who studied more tended to receive higher marks.

### Code explanation

- `plt.scatter(x, y)` creates a scatter plot.
- The first argument supplies x-axis values.
- The second argument supplies y-axis values.
- `plt.grid(alpha=0.3)` adds guide lines to both axes.
- The remaining lines add the title, labels, spacing, and display.

### Important warning

A relationship does not prove that one variable caused the other. Other factors may also affect exam marks.

# 9. Line chart: Monthly shop sales

A line chart shows change across an ordered sequence, especially time.

```python
sales_data = pd.DataFrame({
    "Month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun"],
    "Sales": [120, 135, 128, 150, 165, 180]
})

display(sales_data)
```

### Code explanation

- Each row represents one month.
- `Month` is an ordered time variable.
- `Sales` gives the number of products sold in that month.
- The month order is important. Rearranging it would make the trend misleading.

```python
plt.figure(figsize=(8, 5))
plt.plot(
    sales_data["Month"],
    sales_data["Sales"],
    marker="o"
)
plt.title("Monthly Shop Sales")
plt.xlabel("Month")
plt.ylabel("Products sold")
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

### How to read this line chart

- **X-axis:** Months in time order.
- **Y-axis:** Number of products sold.
- **One marker:** Sales for one month.
- **One line segment:** The change between two consecutive months.
- An upward segment means sales increased.
- A downward segment means sales decreased.
- Sales decreased slightly from February to March and then increased.

### Code explanation

- `plt.plot(x, y)` creates a line graph.
- The first column controls horizontal positions.
- The second column controls vertical positions.
- `marker="o"` places a circular marker at each data point.
- The lines connect data points in the order in which they appear in the table.

### When to use it

Use a line chart for months, years, days, ordered measurements, or another sequence where order matters.

# 10. Multiple-line chart: Sales of two products

A multiple-line chart compares trends measured over the same ordered x-axis.

```python
product_sales = pd.DataFrame({
    "Month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun"],
    "Product_A": [50, 55, 53, 62, 70, 78],
    "Product_B": [40, 46, 52, 57, 60, 68]
})

display(product_sales)
```

### Code explanation

- Each row represents one month.
- `Product_A` and `Product_B` are two separate numerical series.
- Both series are measured using the same unit: number of products sold.
- Because the units and time periods match, comparing them on one chart is reasonable.

```python
plt.figure(figsize=(8, 5))

plt.plot(
    product_sales["Month"],
    product_sales["Product_A"],
    marker="o",
    label="Product A"
)

plt.plot(
    product_sales["Month"],
    product_sales["Product_B"],
    marker="o",
    label="Product B"
)

plt.title("Monthly Sales of Two Products")
plt.xlabel("Month")
plt.ylabel("Products sold")
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

### How to read this multiple-line chart

- **X-axis:** Month.
- **Y-axis:** Products sold.
- **One marker:** One product's sales in one month.
- **One line:** The trend for one product.
- The legend identifies which line belongs to Product A and Product B.
- When one line is above the other, that product had higher sales during that month.
- Both products show a generally upward trend.

### Code explanation

- The first `plt.plot(...)` creates the Product A line.
- The second `plt.plot(...)` creates the Product B line.
- `label="Product A"` and `label="Product B"` give names to the lines.
- `plt.legend()` displays those names in a legend.
- Blank lines in code are optional; they improve readability.

# 11. Area chart: Monthly savings

An area chart is similar to a line chart, but the area under the line is filled.

It can emphasise the size of a value over an ordered sequence.

```python
savings_data = pd.DataFrame({
    "Month_Number": [1, 2, 3, 4, 5, 6],
    "Savings": [1000, 1800, 2500, 3100, 4200, 5000]
})

display(savings_data)
```

### Code explanation

- `Month_Number` uses numbers rather than month names.
- `Savings` contains the amount saved during each month.
- Every row represents one month and its corresponding savings value.

```python
plt.figure(figsize=(8, 5))
plt.fill_between(
    savings_data["Month_Number"],
    savings_data["Savings"],
    alpha=0.4
)
plt.plot(
    savings_data["Month_Number"],
    savings_data["Savings"],
    marker="o"
)
plt.title("Monthly Savings")
plt.xlabel("Month number")
plt.ylabel("Savings amount")
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

### How to read this area chart

- **X-axis:** Month number.
- **Y-axis:** Savings amount.
- **One marker:** Savings in one month.
- **The line:** Change in savings across months.
- **The filled area:** Visually emphasises the magnitude below the line.
- The graph shows that savings increased each month.

### Code explanation

- `plt.fill_between(x, y)` fills the area between the data line and the x-axis.
- `alpha=0.4` makes the filled area partly transparent.
- `plt.plot(...)` draws the line and markers over the filled area.
- Both functions use the same x and y values.

# 12. Box plot: Scores in two classes

A box plot summarises the distribution of numerical data.

It shows the median, quartiles, spread, and possible outliers.

```python
class_a_scores = [55, 60, 62, 65, 67, 70, 72, 74, 78, 82]
class_b_scores = [45, 52, 58, 63, 68, 71, 76, 80, 88, 96]

box_data = pd.DataFrame({
    "Class_A": class_a_scores,
    "Class_B": class_b_scores
})

display(box_data)
```

### Code explanation

- Two lists store scores from Class A and Class B.
- Each score is one student's result.
- The DataFrame places the two classes in separate columns.
- Because both lists have ten values, the table has ten rows.

```python
plt.figure(figsize=(8, 5))
plt.boxplot(
    [class_a_scores, class_b_scores],
    tick_labels=["Class A", "Class B"]
)
plt.title("Score Distribution in Two Classes")
plt.xlabel("Class")
plt.ylabel("Exam score")
plt.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.show()
```

### How to read this box plot

- **X-axis:** The two classes.
- **Y-axis:** Exam score.
- **One box:** The middle 50% of scores in one class.
- **Line inside the box:** The median, or middle score.
- **Lower edge of the box:** First quartile.
- **Upper edge of the box:** Third quartile.
- **Whiskers:** Extend toward lower and higher non-outlier values.
- **Individual dots beyond whiskers:** Possible outliers, when present.
- A taller box indicates greater spread in the middle half of the data.
- Comparing the median lines helps compare the typical score of the two classes.

### Code explanation

- `plt.boxplot(...)` creates a box plot.
- `[class_a_scores, class_b_scores]` is a list containing two lists.
- Each inner list becomes one box.
- `tick_labels=[...]` assigns names to the boxes.
- The y-axis carries the numerical measurement because box plots summarise numerical values.

# 13. Stacked bar chart: Pass and fail counts

A stacked bar chart shows both a total and how that total is divided into parts.

```python
result_data = pd.DataFrame({
    "Class": ["Class 1", "Class 2", "Class 3", "Class 4"],
    "Passed": [28, 32, 25, 30],
    "Failed": [7, 3, 10, 5]
})

display(result_data)
```

### Code explanation

- Each row represents one class.
- `Passed` gives the number of students who passed.
- `Failed` gives the number who failed.
- For each class, passed and failed students together form the total number of students.

```python
plt.figure(figsize=(8, 5))

plt.bar(
    result_data["Class"],
    result_data["Passed"],
    label="Passed"
)

plt.bar(
    result_data["Class"],
    result_data["Failed"],
    bottom=result_data["Passed"],
    label="Failed"
)

plt.title("Pass and Fail Counts by Class")
plt.xlabel("Class")
plt.ylabel("Number of students")
plt.legend()
plt.tight_layout()
plt.show()
```

### How to read this stacked bar chart

- **X-axis:** Class names.
- **Y-axis:** Number of students.
- **One complete bar:** Total students in one class.
- **Lower section:** Students who passed.
- **Upper section:** Students who failed.
- **Total bar height:** Passed plus failed.
- The sections show composition, while the full height shows the total.

### Code explanation

- The first `plt.bar(...)` draws the passed section from zero upward.
- The second `plt.bar(...)` draws the failed section.
- `bottom=result_data["Passed"]` tells Python to begin the failed section at the top of the passed section.
- Without `bottom`, the two bars would overlap instead of stacking.
- `label` names each section.
- `plt.legend()` explains the sections.

# 14. Correlation heatmap: Relationships among numerical variables

A correlation heatmap displays correlation coefficients between numerical variables.

Correlation values usually range from `-1` to `+1`:

- Near `+1`: Strong positive relationship
- Near `0`: Weak or no linear relationship
- Near `-1`: Strong negative relationship

```python
student_data = pd.DataFrame({
    "Study_Hours": [1, 2, 3, 4, 5, 6, 7, 8],
    "Attendance": [55, 60, 66, 70, 78, 82, 88, 94],
    "Sleep_Hours": [5.0, 5.5, 6.0, 6.5, 7.0, 7.0, 7.5, 8.0],
    "Exam_Mark": [40, 46, 53, 59, 68, 74, 82, 90]
})

display(student_data)

correlation_matrix = student_data.corr()
display(correlation_matrix.round(2))
```

### Code explanation

- Each row represents one student.
- The four columns contain numerical variables.
- `student_data.corr()` calculates the correlation between every pair of numerical columns.
- The result is stored in `correlation_matrix`.
- `.round(2)` displays values rounded to two decimal places.
- The same variable compared with itself always has a correlation of `1.00`.

```python
plt.figure(figsize=(8, 6))
plt.imshow(correlation_matrix)
plt.colorbar(label="Correlation coefficient")

plt.xticks(
    range(len(correlation_matrix.columns)),
    correlation_matrix.columns,
    rotation=45,
    ha="right"
)

plt.yticks(
    range(len(correlation_matrix.columns)),
    correlation_matrix.columns
)

for row in range(len(correlation_matrix.columns)):
    for column in range(len(correlation_matrix.columns)):
        value = correlation_matrix.iloc[row, column]
        plt.text(column, row, f"{value:.2f}", ha="center", va="center")

plt.title("Correlation Heatmap")
plt.tight_layout()
plt.show()
```

### How to read this heatmap

- **X-axis:** Numerical variable names.
- **Y-axis:** The same variable names.
- **One cell:** Correlation between the variable in that column and the variable in that row.
- **Cell number:** The correlation coefficient.
- **Colour intensity:** A visual representation of the coefficient.
- The diagonal cells equal `1.00` because each variable is perfectly correlated with itself.
- The matrix is symmetrical. The relationship between A and B equals the relationship between B and A.

### Line-by-line code explanation

- `plt.imshow(correlation_matrix)` displays the matrix as an image.
- `plt.colorbar(...)` adds a scale explaining how colours correspond to values.
- `len(correlation_matrix.columns)` counts the number of columns.
- `range(...)` creates the positions `0, 1, 2, ...`.
- `plt.xticks(...)` sets x-axis tick positions and labels.
- `rotation=45` rotates x-axis labels by 45 degrees.
- `ha="right"` means horizontal alignment is to the right.
- `plt.yticks(...)` sets y-axis tick positions and labels.
- The two `for` loops visit every row and every column.
- Indentation shows which statements belong inside the loops.
- `correlation_matrix.iloc[row, column]` retrieves one matrix value by numerical position.
- `f"{value:.2f}"` is an f-string:
  - `f` allows a variable to be inserted into text.
  - `{value:.2f}` displays the value with two decimal places.
- `plt.text(...)` writes the number in the correct heatmap cell.
- `ha="center"` and `va="center"` centre the text horizontally and vertically.

### Important warning

Correlation does not prove causation. A high correlation only means that two variables move together in this dataset.

# 15. Choosing the correct graph

| Purpose | Recommended graph |
|---|---|
| Show the distribution of one numerical variable | Histogram |
| Compare separate categories | Bar chart |
| Compare categories with long names | Horizontal bar chart |
| Show parts of one total | Pie chart |
| Study the relationship between two numerical variables | Scatter plot |
| Show change over time | Line chart |
| Emphasise magnitude over an ordered sequence | Area chart |
| Compare median, spread, and outliers | Box plot |
| Show totals divided into components | Stacked bar chart |
| Compare correlations among numerical variables | Correlation heatmap |

# 16. Common interpretation mistakes

1. **Calling every plotted object a data point**  
   A scatter-plot dot can represent one observation, but a histogram bar represents a range containing several observations.

2. **Using a line chart for unordered categories**  
   Connecting unrelated categories may falsely suggest continuity or change.

3. **Using a pie chart when values do not form one total**  
   Pie slices should represent parts of a meaningful whole.

4. **Ignoring axis labels and units**  
   A number is difficult to interpret without knowing whether it means students, marks, hours, money, or percentages.

5. **Assuming correlation proves causation**  
   Two variables moving together does not prove that one causes the other.

6. **Starting a bar-chart y-axis far above zero without explanation**  
   This can exaggerate small differences.

7. **Using too many categories**  
   Too many bars, slices, or lines can make a graph difficult to read.
